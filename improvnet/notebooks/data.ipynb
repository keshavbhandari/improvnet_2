```python
import os
import requests

# 1️⃣ Your Hugging Face token
HF_TOKEN = "YOUR_HF_TOKEN"  # replace with your token

# 2️⃣ URL of the file
url = "https://huggingface.co/datasets/Metacreation/GigaMIDI/resolve/main/Final_GigaMIDI_V1.1_Final.zip"

# 3️⃣ Where to save it
save_path = "/data/improvnet/GigaMIDI/Final_GigaMIDI_V1.1_Final.zip"
os.makedirs(os.path.dirname(save_path), exist_ok=True)

# 4️⃣ Create headers with token for authentication
headers = {"Authorization": f"Bearer {HF_TOKEN}"}

# 5️⃣ Download file
with requests.get(url, headers=headers, stream=True) as r:
    r.raise_for_status()  # raise error if something goes wrong
    with open(save_path, "wb") as f:
        for chunk in r.iter_content(chunk_size=8192):
            f.write(chunk)

print(f"Downloaded successfully to {save_path}")
```
